```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
nsteps=100000

def euler_newton(force, times, x0, v0, m,dt,):

    x = [x0]
    v = [v0]
    b = 0.2 
    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v.append(v[i] + a*dt)
        x.append(x[i] + v[i]*dt)
    return x, v
times,dt = np.linspace(0, 10, nsteps,retstep=True)

def gravity(t, x, v):
    output = -m*g
    return output
    
def f(times, x0, v0, g):
    output = x0 + v0*times - (1/2)*g*times**2
    return output
g = 9.8
x0 = 10
v0 = 0
m = 5
print(gravity(0, 0, 0))

print(f(times, x0, v0, g))

def lin_drag(times, x0, v0):
    output = -m*g - b*v0
    return output
m = 10 #kg
x0 = 0 #m 
v0 = -50 #m/s
print(lin_drag(times, x0, v0))

def quad_drag(times, x0, v0):
    output = -m*g -c*v0*np.abs(v0)
    return output 
m = 10 #kg
c = 2 #kg/m
x0 = 0 #m 
v0 = -50 #m/s
print(quad_drag(times, x0, v0))

x,v = euler_newton(gravity,times,x0,v0,m,dt)
v
euler_newton(lin_drag, times, x0, v0, m, dt)

euler_newton(quad_drag, times, x0, v0, m, dt)
```

```python
def lin_drag(times, x0, v0):
    output = -m*g - b*v0
    return output
m = 10 #kg
b = 5 #kg/s
x0 = 0 #m 
v0 = -50 #m/s
print(lin_drag(times, x0, v0))
```

```python
x, v = euler_newton(lin_drag, times, x0, v0, m, dt)
plt.plot(times, v)
plt.plot((0, 10), (-19.6, -19.6), linestyle = "dashed")
plt.grid()
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.tight_layout()
plt.show()

plt.plot(times, x)
plt.grid()
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.tight_layout()
plt.show()
```

```python
def quad_drag(times, x0, v0):
    output = -m*g -c*v0*np.abs(v0)
    return output 
m = 10 #kg
c = 2 #kg/m
x0 = 0 #m 
v0 = -50 #m/s
print(quad_drag(times, x0, v0))
```
