```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
nsteps=100000

def euler_newton(force, times, x0, v0, m,dt):

    x = [x0]
    v = [v0]
    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v.append(v[i] + a*dt)
        x.append(x[i] + v[i]*dt)
    return x, v
times,dt = np.linspace(0, 10, nsteps,retstep=True)

def gravity(t, x, v):
    output = -m*g
    return output
    
def f(times, x0, v0, g):
    output = x0 + v0*times - (1/2)*g*times**2
    return output
g = 9.8
x0 = 10
v0 = 0
m = 5
print(gravity(0, 0, 0))

print(f(times, x0, v0, g))

x,v = euler_newton(gravity,times,x0,v0,m,dt)
v
```

```python
vt=v0-g*times
xt=x0+v0*times-.5*g*times**2
plt.plot(times, v)
plt.plot(times, vt,linestyle="dashed")
plt.grid()
plt.xlabel("time")
plt.ylabel("velocity")
plt.tight_layout()
plt.show()

plt.plot(times, x)
plt.plot(times, xt,linestyle="dashed")
plt.grid()
plt.xlabel("time")
plt.ylabel("position")
plt.tight_layout()
plt.show()
```
